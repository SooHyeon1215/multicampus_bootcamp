### 자연어 처리
- NLP의 접근 방식
    - 규칙 기반 : 사람의 규칙(문법/패턴)을 정의
    - 통계 기반 : 단어의 빈도, 확률, 통계를 활용
    - 딥러닝 기반 : 대규모 데이터 + 신경망 모델
- N-Gram
    - Unigram
        - N = 1
        - 독립 단어
        - 단어를 하나씩 가지고 벡터화
    - Bigram
        - N = 2
        - 바로 앞의 단어를 고려
        - 2개의 단어를 이용하여 벡터화
    - Trigram
        - N = 3
        - 앞의 2개의 단어를 고려
        - 3개의 단어를 이용하여 벡터화
- 로그 확률
    - 조건부의 확률이 매우 작은 값인 경우(0.0001)
    - 이런 확률끼리 곱을 하게 되면 -> 0에 가까운 값으로 생성 -> 계산이 불안정
    - 이러한 문제를 해결하기 위해서 log 값을 사용
- 토큰화
    - 문장을 잘게 쪼개는 방법
    - 영문이라면 split() 함수를 이용해서 나눠주는 방식을 사용
    - 한글의 경우에는 해당 방식으로는 문제가 발생 할 수 있다 (의미를 명확하게 파악이 힘들다)
    - 라이브러리의 종류 : konlpy, sentencepiece
        - konlpy
            - 기존에 단어 사전이 존재하는 객체(class)
            - Okt
                - SNS, 댓글등의 한글로 이루어진 데이터를 형태소 분석하기 위한 분석기
                - 품사 태그가 kkma, komoran과는 다른 품사 태그를 사용
            - kkma
                - 굉장히 깐깐한 언어 분석기
                - 국어 사전에 나온 단어들을 이용하여 단어를 토큰화
                - 분석의 시간이 오래 걸린다는 단점
                - OOV 문제가 발생
            - Komoran
                - 기본적인 구조는 kkma와 같은 단어장을 기준으로 문장을 토큰화
                - 기본의 단어 사전에서 새로운 단어들을 보강하여 토큰화 작업이 가능
            - 품사 태그
                - Okt
                    - Noun : 명사
                    - Verb : 동사
                    - Adjective : 형용사 = ....
                - kkma, komoran
                    - 세종 태그셋을 사용
                    - 감성 / 의도 분석
                        - NNG(일반 명사), NNP(고유 명사), VV(동사), VA(형용사), MAG(일반 부사), SL(외국어)
        - sentencepiece
            - 문장들의 덩어리를 학습하여 글자 간의 상호 작용을 통해서 단어 사전을 생성
            - Train()을 이용하여 model이 생성이 되고 vocab이 생성
            - 딥러닝 기반 토큰화에서 자주 사용이 됨 (특정 모델의 토큰화 방식들은 일반적으로 sentencepeice 구조로 되어있다.)
- 통계 기반
    - 통계 기반 예측 모데을 만드는 순서
        - 데이터를 로드 -> 문장들을 토큰화 -> 토큰화 및 데이터를 벡터화 -> 모델에 학습 -> 예측
    - BoW (Bag of Word)
        - 각 문서에서 단어가 몇번 등장했는가? (빈도수)
        - 문서에서 단어의 순서는 무시하고 등장 횟수로 의미를 찾는 방법
        - 가장 기본적인 벡터화 방법
        - 모델을 이용해서 예측
    - TF-IDF
        - BoW 벡터화의 한계를 보완하기 위한 통계 기반 기법
        - TF
            - 특정 문장 내에서 단어들의 빈도수가 클수록 값이 큼
            - 자주 등장할 수록 해당 문서에서 중요하다고 판단
        - IDF : 전체 문서에서 단어들의 비율의 역수
        - 단순히 단어의 개수를 이용하는것이 아니라 특정 문서에서 유독 자주 나오지만 전체 문서에서 잘 등장하지 않는 단어들에게 가중치를 부여하는 방법
        - TF * IDF로 계산되어 실수형 벡터화
        - 매개변수
            - min_df
                - 최소 문서의 빈도수
                - 너무 희귀한 단어를 제외시킨다.
            - max_df
                - 최대 문서의 빈도수
                - 너무 자주 나온 단어를 제외
            - stop_words : 불용어 처리
    - 코사인 유사도
        - 문장과 문장 사이에 어느 정도 같은 의미를 가지는가?
        - 유클리드 거리법은 단순 거리 계산 방식은 고차원이 되면 거리가 멀어지는 현상이 발생
        - 거리가 아닌 각도를 기준으로 유사도를 체크하는 방법
        - 가장 일반적으로 사용하는 방법
        - 1 : 같은 의미
        - 0 : 서로 독립적
        - -1 : 상반된 의미
-딥러닝 기반
    - Word2Vec
        -문자를 숫자 데이터로 변환 시켜주는 답러닝 기반의 임베딩 기술
            - CBOW : 주변의 단어들을 이용하여 중심 단어를 예측
            - Skip-gram : 중심 단어를 이용하여 주변의 단어들을 예측
            - 빠른 계산을 원하면 CBOW를 사용한다.
            - 일반적으로 Skip-gram을 사용
            - sg 매개변수를 이용해서 설정
        - 객체를 생성하고 학습한 뒤 wv 속성의 값은 class를 되돌려준다.
        - CBOW, Skip_gram을 혼합하여 사용하기 한다.
    - FastText
        - Word2Vec에서 OOV문제를 해결하기 위한 모델
        - Word2Vec은 단어를 기준으로 예측을 한다면 FastTest는 글자를 기준으로 예측 모형
        - '강아지' -> 강, 아, 지 나눠서 subword를 생성
        - Word2Vec 에서 강아지와 강아지들을 다른 단어로 인식을 한다면 FastText 유사한 단어로 인식
    - Doc2Vec
        - Word2Vec 확장판
        - 기본적인 Word2Vec의 학습 내용은 그대로 유지 , 문장의 벡터를 생성
        - 속성의 종류
            - 단어 벡터의 저장소 (wv)
            - 문장 벡터의 저장소 (dv)
        - 문장의 id 값들을 지정하여 모델에서 학습 (TaggedDocument)
    - TextCNN
        - CNN(합성곱 신경망) : 이미지 분석시 사용하는 신경망 모델
        - 텍스트를 이미지와 같다 라고 가정을 하고 설계한 모델
        - 단어의 흐름 방향으로 구간을 선택하여 데이터를 학습
        - 구간별 데이터들을 합성곱을 통해 가장 큰 값을 선택
        - 해당 데이터들을 이용하여 모델의 독립 변수로 사용
        - 전체 문장에서 특정 한 구간이 해당 문장의 의미를 가진다
        - n-gram를 활용하여 구간을 생성하는데 3, 4, 5로 피쳐의 개수가 n의 개수만큼 증가하기 때문에 과적합의 위험성 존재 ( dropout을 통해서 일부의 데이터를 소실 )
    - RNN, LSTM
        - 단어들의 순서들 이용하여 벡터화
        - RNN은 문장이 길어질수록 과거의 데이터의 가중치가 소실이 되는 문제가 발생
        - LSTM은 게이트를 이용해서 망각, 장기 기억을 나눠서 작업 (기울기 소실 문제를 어느 정도 방지)
        - 토큰화 된 데이터를 embedding을 통해서 벡터화를 하고 RNN, LSTM 모델을 이용하여 학습 및 예측
        - 비선형 활성화 함수 (ReLU, tanh)를 사용하지 않는다. (RNN, LSTM 안에 존재)
    - transformer
        - RNN, LSTM 모델에서는 문장을 순차적으로 처리하지만 문장 전체를 한번에 병렬로 처리하면서 단어 간의 관계를 파악 (self-attention)
        - 문장의 순서보다는 단어들 간의 관계를 중요하게 생각
        - self-attention
            - 3개의 단어 벡터를 생성 (Q(질문자), K(색인), V(전체 정보))
            - Q, K를 내적(유사도 계산) -> softmax()를 이용하여 확률 생성 -> V와 곱셈(가중치 곱셈)
        - attetion의 결과를 바탕으로 비선형 변환
        - 선형 모델을 이용해서 예측
        - tranformer 모델을 대규모의 데이터를 이용하여 학습을 해야 준수한 성능이 만들어진다.
        - 허깅 페이스를 이용하여 미리 학습이 된 모델을 로드하여 사용
            - 미리 학습된 모델 -> 유치원 정도의 지식을 가진 모델(가장 일반적인 지식만 가지고 있는 상태)
            - 파인튜닝 -> 추가 학습을 통해서 모델의 성능을 내가 가지고 있는 데이터의 형태로 모델의 성향을 변경 (고급화 과정)
        - encoder : 입력된 데이터를 이해
        - decoder : 텍스트를 생성하는 과정
        - kobert
            - encoder only 모델
            - 문장을 이해하는 방식은 문장에서 단어들을 랜덤하게 마스킹 처리하여 해당 단어들을 예측하는 과정 
            - 종속 변수는 0/1과 같은 답을 예측을 하는 모델
        - kobart
            - encoder + decoder 혼합 모델
            - 문장을 이해하는데는 BERT 모델과 같은 방식을 이용
            - decoder에서는 하나의 문장으로 통째로 마스킹 처리하여 텍스트 생성하면서 학습
            - generate()를 통해서 예측 값이 텍스트로 생성
            - 요약본, 질문에 대한 답변, 분류 모델로도 사용은 가능

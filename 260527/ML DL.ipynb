# 머신러닝
- 종류
    - 지도학습
        - 선형 모델
            - 데이터는 모두 선으로 예측이 가능하다 (추론에서부터 시작)
            - 입력 특성(독립변수)에 따라 선형 함수를 만들어 예측하는 알고리즘
        - 트리 모델
            - 조건식을 사용하여 데이터를 분할하고 불순도가 낮을 떄까지 계속적으로 데이터를 분할하여 예측하는 방법
            - 불순도 : 지니계수, 엔트로피, log_loss
    - 비지도학습 : 정답이 없는 데이터셋을 이용
        - 차원 축소
            - PCA, t-SNE
            - feature가 개수가 너무 많은 경우 데이터의 계산에서 시간이 너무 많이 소모가 되기 때문에 시간을 줄이기 위해서 차원을 축소
            - 다차원의 데이터를 그래프 시각화 하기에 어려움이 존재하기 때문에 차원 축소를 이용하여 그래프 시각화
            - 텍스트 데이터를 토큰화한 뒤 단어 간의 유사성을 보여주기 위해서도 사용함
        - 군집화
            - 비슷한 유형의 데이터들을 모아서 하나의 그룹으로 생성
            - 군집의 개수를 지정할 수 있는 K-Means
                - 밀도 기반의 군집화
                - 고밀도 지역의 데이터들을 하나씩 군집으로 생성
                - 경계의 데이터들이 군집을 형성할 수 있는가? 확인
                - 형성이 가능하면 군집의 크기를 늘려준다.
                - 군집에 포함되어있지 않은 데이터는 노이즈로 판단

### 지도학습 모델의 종류
1. 선형 회귀 모델 (LinearRegression)
    - 선형 모델의 기본 모델
    - 독립 변수들의 feature들 간의 선형 함수를 생성하여 결과를 예측하는 모델
    - 비선형 데이터를 예측하려면 성능이 좋지 않다.
    - 비선형 데이터를 예측하기 위해선 다항 회귀 (PolynomialFeatures)
        - feature의 개수를 늘려준다. (상호 관계에 따른 feature, 제곱합, 절편합)
3. 가중치에 패널티를 줌으로써 예측도를 올려주는 모델
    - 릿지 (Ridge)
        - L2 패널티를 이용하여 가중치에 변화를 줌으로써 모델의 성능을 올려준다.
        - 패널티의 강도가 클수록 회귀계수가 적어지고 과적합을 방지, 과소적합의 위험 증가
    - 라쏘 (Lasso)
        - L1 패널티를 이용하여 가중치에 변화를 주는 모델
        - 가중치의 변화가 규제 강도에 따라 0까지 감소
            - 컬럼의 개수가 매우 많은 경우 컬럼을 선택할 수 있도록 도와준다.(feature selection)
            - 모델 단순화의 가능성 존재
    - 엘라스틱넷 (ElasticNet)
        - 릿지, 라쏘 둘을 절충한 모델
        - 두개의 패널티를 혼합하여 모델의 성능을 올린다.
        - l1_ratio 매개변수를 이용하여 두 패널티의 비율을 설정
3. 로지스틱 회귀 (LogisticRegression)
    - 이름은 회귀지만 결과는 분류 모델
    - 0과 1사이의 값(실수)을 예측하여 0.5보다 큰 경우라면 1, 아니라면 0을 출력하는 알고리즘
    - 종속변수가 범주형인 경우 적용하는 회귀 분석
    - 종속변수 y를 직접 모델링에 활용하지 않고 y가 특정 범주에 속하는 확률을 모델링
    - 분류형 모델들은 데이터의 불균형에 대한 매개변수 class_weight 존재
4. 서포트벡터머신 (SVM)
    - 데이터를 분리하는 최적의 경계(결정 경계)를 찾고 해당 경계선에서 어느정도의 마진(여유공간) 생성
    - 회귀의 경우에는 마진 밖에 데이터들을 공간 안으로 들어올 수 있도록 패널티 부여
    - 분류의 경우에는 마진 영역 안쪽의 데이터들을 공간 밖으로 빠질 수 있도록 패널티 부여
5. 의사결정나무 (DecisionTree)
    - 나무 구조를 도표화하여 분류와 예측을 수행하는 분석 방법
    - 적절한 규칙(조건식)을 찾아서 데이터 분할하여 나무를 성장시키는 과정
    - 불순도를 측정하여 불순도가 낮은 방향으로 자식 가지를 형성
    - 나무의 깊이가 깊어질수록 불순도를 내려가지만 과적합의 위험성이 존재
6. 앙상블 (ensemble)
    - 단일 결정 트리의 단점을 극복하고자 여러 ML 모델을 연결하여 더욱 강력한 모델을 생성
    - 대표적인 방법 : 배깅, 부스팅
        1. 배깅
            - 대표적인 모델 : 랜덤 포레스트
            - 부트스트랩 자료를 생성 (전체 데이터에서 일부의 데이터를 가져와서 모델에 입력)
            - 여러 모델들을 각각 독립적으로 학습을 하고 결과를 다수결의 원칙에 따라서 다수의 의견을 최종 결과물로 만들어준다.
            - 부트스트랩 자료, 컬럼의 선택을 하는 이유는?
                - 각 모델들에 데이터의 다양성을 가질 수 있도록 하기위함
            - 장점 : 모델의 불안정성을 줄이고 분산이 감소
            - 단점 : 편향은 그대로 유지
        2. 부스팅
            - 대표적인 모델 : XGBoost, LightGBM, CatBoost
            - 여러개의 모델들을 직렬화하여 하나의 모델의 결과를 이용하여 오차를 계산하고 오차를 줄이기 위해 다음 모델을 사용하는 형태
            - 장점 : 편향 감소
            - 단점 : 순차적인 학습을 통해 학습 시간이 증가, 학습의 횟수가 너무 많은 경우 과적합, 매개변수 튜닝이 매우 중요

### 교차 검증
1. K-Fold
    - 데이터를 k개의 동일한 크기로 나눠준다. (폴드화)
    - k번 반복하여 하나의 폴드는 검증용으로 나머지 폴드는 학습용으로 시용
    - 모든 폴드는 한번씩 검증용으로 사용
    - 데이터의 수가 적은 경우에는 모델의 성능을 안정적으로 평가할 수 있는 방법
    - 계층화(Strarified) K-Fold 존재
2. GridSearchCV
    - 매개변수의 조합을 생성하고 최적의 조합의 값을 찾는 방법
    - 조합별로 모델을 학습하고 검증하여 평가 지표를 통해서 좋은 모델의 매개변수의 조합을 확인
3. Pipeline
    - 머신러닝의 학습 과정을 하나씩 연속된 흐름으로 묶는 도구
    - 데이터의 전처리 과정, 모델의 학습, 예측과 같은 흐름을 하나로 연결하는 기능
    - 장점 : 코드 간결화 및 자동화, 데이터의 누수 방지, 재사용성이 높음
    - 단점 : 오류 추적이 문제, 학습 시간이 증가, 순차적인 진행이기 떄문에 병렬화의 어려움

### 평가 지표
    - 회귀 모델
        - MAE 
            - 실제 값과 예측값의 차이를 절대값으로 변환하여 평균을 구한 값
        - MSE 
            - 실제값과 예측값의 차이를 제곱하여 평균을 구한 값
            - 실제값과 예측값의 면적을 의미
        - RMSE 
            - MSE값에 루트를 씌운 값 
            - 오차를 젝곱하면 오차가 클수록 기하급수적으로 증가하는 상황이 발생하기 때문에 해당 현상을 상쇄하기 위해 시용
        - MSLE
            - MSE값에 로그를 취한 값
            - RMSE와 같이 기하급수적으로 커지는 상황을 방지하기 위해 사용
        - MAPE : MAE를 퍼센트로 표시
        - R2_Score
            - 1 - (SSE / SST)
                    - SSE : 설명하지 못하고 남은 에러
                    - SST : 전체 분산
                    - 1에 가까울수록 모델의 성능이 좋다.
    - 분류 모델
        - 혼동 행렬
            - 분류 모델에서 예측 오류가 얼마인가? 어떠한 유형의 오류가 발생하는? 를 나타내는 지표
                - TN : 예측이 부정, 실제가 부정
                - FP : 예측이 긍정, 실제가 부정
                - FN : 예측이 부정, 실제가 긍정
                - TP : 예측이 긍정, 실제가 긍정
        - 정확도
            - 실제의 데이터에서 예측 데이터가 얼마나 같은가 판단
            - TN + TP / TN + FN + FP + TP
        - 정밀도 
            - 긍정으로 예측한 것들 중 실제가 긍정인 비율
            - TP / TP + FP
        - 재현율
            - 실제 긍정인 것들 중 예측이 긍정인 비율
            - TP / TP + FN
        - F1_Score
            - 정밀도와 재현율을 결합한 분류 성능 지표
            - 2 * TP / 2 * TP + FP + FN

### 데이터 전처리
1. 이상치
    - 값이 이상한 데이터
        - 결측치
        - 범위에서 벗어나는 데이터
        - 값이 크게 벗어나는 데이터 (극단치)
    - 분석 모델의 성능을 떨어뜨리거나 결과에 악영향을 미치는 경우가 발생
    - 이상치 데이터는 즉시 처리
        - 데이터 제거
            - 소량의 데이터가 이상치인 경우 제거 (일반적으로 전체 데이터에서 5% 미만인 경우)
            - 학습 데이터가 사라진다는 문제가 발생 (다량의 데이터가 이상치로 판단이 되면 다른 값들로 대체)
        - 특정 값 대체
            - 일반적으로 평균 값이나 최빈값으로 대체
            - 데이터의 신뢰도가 떨어질 수 있다.
            - 분석 성능이 떨어지는 경우 발생 (데이터의 분포가 과밀화되는 경우)
    - 결측치의 확인
        - isna() 함수를 이용해서 결측치의 유무 판단 -> sum() 함수를 이용해서 결측치의 개수를 확인
        - isna() 함수를 이용하고 any() 함수를 활용하면 결측치가 포함되어있는 행이나 열 확인
    - 특정 범위에서 벗어나는 데이터 확인
        - 특정 열에 데이터의 유일 값을 출력 : unique()
        - 특정 열에 데이터의 빈도를 확인 : value_counts()
        - Series 데이터를 집합의 형태로 변경 : set()
    - 극단치를 확인하는 방법
        - IQR 방식
            - boxplot()의 수염의 경계
            - IQR = Q3 - Q1
            - 극단치의 경계
                - 상단의 경계 : Q3 + 1.5 * IQR
                - 하단의 경계 : Q1 - 1.5 * IQR
            - 상단의 경계보다 크거나 하단의 크기보다 작은 값이 극단치
2. 데이터의 불균형 문제
    - 정상을 정확하게 분류하는 것과 이상을 정확하게 분류하는 것 중 이상을 정확하게 분류하는 것이 중요
        - 예) 신체 정보를 이용하여 건강의 이상을 탐지한다 -> 건강한 사람도 이상이라고 감지하여 병원에 가도록 하는 것이 중요
    - 일반적으로 데이터들은 정상의 데이터가 이상의 데이터보다 많다.
    - 해당 데이터를 학습으로 모델에 입력하게 되면 다수의 데이터를 더 학습하여 다수의 라벨(정답)으로 예측의 확률이 올라간다.
    - 다수의 데이터를 소수의 데이터만큼 줄여주는 방법 (RandomUnderSampling)
        - 데이터가 제거됨으로 학습 데이터가 줄어든다. (학습율이 감소함으로 모델의 성능에 영향)
        - 데이터의 개수가 제거해도 충분히 많은 경우
        - 데이터의 신뢰도가 굉장히 중요한 경우
    - 소수의 데이터를 다수의 데이터만큼 늘리는 방법
        - 단순 복제 (RandomOverSampling)
            - 소수의 데이터를 단순 복제하여 다수의 데이터 개수에 맞춰가는 방법
            - 데이터의 손실이 없기 때문에 학습 데이터는 증가
            - 과적합의 위험성 존재
        - 최근접 데이터 생성 (SMOTE)
            - 소수의 데이터 관측값에 대한 k개의 최근접 양수를 이웃으로 찾고 관측값과 이웃으로 선택된 값 사이에 임의의 데이터를 생성하는 방법
            - 랜덤 오버 샘플링에서 과적합의 위험성을 보완한 샘플링 기법
    - 분류 모델에서 clas_weight 매개변수를 이용해서 소수의 데이터에 가중치 패널티를 부여하는 방법
3. 스플릿 (split : 데이터의 분할)
    - 분석 모델에서 학습 데이터와 검증 데이터로 데이터는 분할
    - 학습 시간 모델의 성능을 펼가하기 위함
    - train, test 셋으로 데이터를 분할 (8:2, 7:3)
    - shuffle 매개변수 : 분할은 처음부터 80%까지 train, 나머지는 test의 구조에서 데이터 분할 전 데이터들의 순서를 섞어주는가?
        - shuffle을 사용하면 안되는 경우 : 시계열 데이터인 경우
    - startify 매개변수 : 특정 열을 선택하면 해당 열의 비율에 맞춰서 데이터를 분할
4. 스케일링
    - 컬럼 간의 데이터의 범위가 크게 차이가 나는 경우 모델이 정상적으로 작동하지 않는다.
        - 범위가 큰 컬럼이 예측에 상대적으로 큰 영향을 준다.
    - 각 컬럼 간의 범위를 맞춰주기 위한 작업 스케일링
    - StandardScaler
        - 평균이 0, 분산이 1인 정규 분포로 스케일링
        - 이상치 민감
    - MinMaxScaler
        - 최솟값을 0, 최댓값을 1로 0과 1 사이의 값을 스케일링
        - 이상치 민감
    - MaxAbsScaler
        - 최대 절댓값과 0이 각각 1, 0이 되도록 스케일링
        - 데이터가 모두 양수인 경우에는 MinMaxScaler와 동일
        - 이상치 민감
    - RobustScaler
        - 중앙값과 사분위수를 활용하는 스케일링 방식
        - 중앙값을 0으로 설정하고 IQR을 사용하여 이상치의 영향을 최소화
        - 위 4가지 스케일러 중에서 가장 이상치에 덜 민감함

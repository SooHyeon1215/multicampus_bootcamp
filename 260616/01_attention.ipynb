```python
import torch
import torch.nn.functional as F
```

```python
#임의의 문장 임베딩 데이터
#3개의 단어에 각 단어는 4개의 차원의 숫자로 표현
x=torch.randn(3, 4)
x
```

```python
#Query, Key, Value를 임의의 가중치 행렬로 생성 (원래는 학습이 된 가중치 행렬 부분)
w_q=torch.randn(4, 4)
w_k=torch.randn(4, 4)
w_v=torch.randn(4, 4)
```

```python
#x의 데이터와 각각의 가중치 행렬을 곱하여 q, k, v를 생성
q=torch.matmul(x, w_q)
k=torch.matmul(x, w_k)
v=torch.matmul(x, w_v)
```

```python
#Attention Score를 생성 -> Query와 Key의 내적
attention_score = torch.matmul(q, k.T)
```

```python
#Attenion_score는 고차원잀록 발산을 하기 때문에 Key차원 수에 루트를  씌워서 나눠준다.
d_k = k.size(1)
```

```python
#스케일링과 softmax를 통해서 확률로 변환 (모든 행의 합이 1)
attention_weight = F.softmax(attention_score / (d_k ** 0.5), dim=1)
attention_weight
```

```python
output = torch.matmul(attention_weight, v)
output
```

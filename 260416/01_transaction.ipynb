```python
#transation, item_master 조인 결합 -> 공통 컬럼은 item_id
df=transaction.merge(
    item_master,
    on='item_id',
    how='inner'
)
```

```python
df.info()
#item_master에서 결합된 컬럼 item_name, item_price의 결측치가 존재하지 않는다.
#같은 집합 : item_id 기준으로 inner 결합 -> 인첵스 유지 / 결측치 존재 X 
```

```python
#df와 customer_master 조인 결합 -> 조건 customer_id
df=pd.merge(
    df, customer_master,
    on='customer_id',
    how='inner'
)
```

```python
#시계얄 데이터 patment_date 컬럼을 타입을 변경
#문자를 시계열로 변경(datetime.strptime()/to_datetime())
pd.to_datetime(df['payment_date'])
```

```python
#datetime 라이브러리의 strptime()을 이용하여 데이터 타입 변환
from datetime import datetime
```

```python
df['payment_date'][0]
```

```python
#datetime안에 strptime() 함수는 to_datetime()과는 다르게 Series형태 변환이 불가능
# datetime.strptime(df['payment_date'][0],'%Y-%m-%d %H:%M:%S')
#Series 데이터에서 value들을 하나씩 뽑아서 함수에 넣는다. map()
df['payment_date'].map(
    #x에 대입되는 데이터는 문자
    lambda x : datetime.strptime(x,'%Y-%m-%d %H:%M:%S')
)
```

```python
#Series 데이터에서 문자열 함수를 사용하기 위해
    #1.반복문 사용
    #2.map()+lambda
    #3.문자열 함수를 사용할 수 있도록 pandas에서 제공한 Series.str (String Method)
#Series에서는 시계열 함수를 사용할 수 있도록 문자열과 같이 DatetimeMethods 변환이 존재
# Series.str --> 해당 데이터셋이 시계열로 시계열 함수
```

```python
df['payment_date']=pd.to_datetime(df['payment_date'])
```

```python
df.info()
```

```python
#요일별 판매가 된 제품들의 총 판매 개수들의 함계가 어떻게 되는가?
#요일이라는 새로운 변수를 생성 -> 시계열에서 요일만 따로 추출(strftime(%A | %a | %w))
df['요일']=df['payment_date'].dt.strftime('%A')
#strftime() : 0(일),...
```

```python
#transaction데이터에서 customer_id와 customer_master 데이터에서 customer_id값들이 다른가?

list_1=transaction['customer_id'].unique()
```

```python
list_2=customer_master['customer_id'].unique()
```

```python
#두개의 리스트에서 서로 다른 값이 다른 데이터는 어떻게 되는가?
new_list=[]

for customer in list_1:
    if customer  not in list_2:
        #고객의 ID가 list_2에 존재하지 않는다면
        new_list.append(customer)

new_list
```

```python
[customer for customer in list_1 if customer not in list_2]
```

### Linear Regression
- parameter(매개변수 -> class 생성시 자동 호출이 되는 생성자 함수의 매개변수)
    - fit_transform
        - 기본값 : True
        - 신형 모델에서 절편을 사용할 것인가?
    - normalize
        - 기본값 : False
        - 정규화를 할 것인가? -> 일반적으로는 Scaler를 이용하여 데이터의 범위를 지정
    - copy
        - 기본값 : True
        - 입력된 독립변수(x)를 복사하여 사용할 것인가?
    - n_jobs
        - 기본값 : None
        - 계산 작업을 하는 CPU의 개수를 지정
    - positive
        - 기본값 : False
        - 회귀 계수가 음수를 사용하지 않도록 설정
- 속성(class에 내장된 변수(data))
    - coef_
        - 선형 회귀에서 추정된 회귀 계수
    - rank_
        - 선형 독립 차원의 수
            - 독립변수(컬럼)의 개수와 rank_의 수가 같다면 모든 변수가 독립젹
            - rank_의 수가 독립변수의 개수보다 작은 경우 다중공선성(컬럼 간의 중복)이 존재
    - singular_
        - 독립 변수의 특이값
    - interect_
        - 절편 값
- 매서드(class에 내장된 함수)
    - fit()
        - 모델을 학습 시키는 함수(독립 변수와 종속 변수 간의 규칙을 찾는 과정)
        - X : 학습 데이터(독립 변수), 2차원 데이터가 입력됨(DataFrame, 2차원 array)
        - y : 정답 데이터(종속 변수), 1차원 데이터(Series, 1차원 array)
        - sample_weight(옵션) : 개별 데이터에 특정 가중치
    - get_params()
        - 선형 회귀 모델을 생성할 때 지정한 매개변수의 값들을 출력
        - X : 평가 데이터(독립 변수)
        - 예측한 값들은 array의 형태로 되돌려준다.
    - score()
        - 예측에 대한 결정 게수를 출력
        - X : 평가 데이터(독립 변수)
        - y : 종속 변수

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
#데이터 로드
boston=pd.read_csv('../csv/boston.csv')
```

```python
boston.head()
```

```python
#데이터 분할(독립, 종속 변수를 train, test로 나눠준다.)
X_train,X_test,y_train,y_test=train_test_split(
    boston.drop('Price',axis=1),
    boston['Price'],
    test_size=0.3,
    random_state=42
)
```

```python
#스케일링 -> train데이터를 이용하여 범위를 지정하고 train, test 모두 스케일링
#객체 생성 -> train 데이터 fit() -> train, test를 transform()
stdScaler=StandardScaler()
```

```python
X_train_sc=stdScaler.fit_transform(X_train)
X_test_sc=stdScaler.transform(X_test)
```

```python
#선형 회귀 모델을 생성
#기본값으로 생성
lr=LinearRegression()
#절편 제외 모델 생성
lr_inter=LinearRegression(fit_intercept=False)
```

```python
#기본형 모델에 스케일링이 되지 않은 데이터를 사용
lr.fit(X_train,y_train)
```

```python
pred=lr.predict(X_test)
```

```python
print(mean_squared_error(y_test,pred))
print(r2_score(y_test,pred))
```

```python
lr.fit(X_train_sc,y_train)
```

```python
pred=lr.predict(X_test_sc)
```

```python
print(mean_squared_error(y_test,pred))
print(r2_score(y_test,pred))
```

```python
#절편을 사용하지 않는 모델을 이용하여 확인
lr_inter.fit(X_train,y_train)
```

```python
pred_inter=lr_inter.predict(X_test)
```

```python
print(mean_squared_error(y_test,pred_inter))
print(r2_score(y_test,pred))
```

```python
lr_inter.fit(X_train,y_train)
```

```python
pred_inter=lr_inter.predict(X_test_sc)
```

```python
print(mean_squared_error(y_test,pred_inter))
print(r2_score(y_test,pred_inter))
```

```python
#절편을 사욯한 모델과 절편을 사용하지 않은 모델의 회귀 계수를 확인
print(lr.coef_)
print(lr_inter.coef_)
```

```python
#score() 함수 -> 결정 계수
lr.score(X_train_sc,y_train)
```

```python
lr_inter.score(X_train_sc,y_train)
```

#### Okt
- SNS, 댓글 등의 한글로 이루어진 데이터를 형태소 분석하기 위한 분석기
- 주요 품사 태그 종류
    - Noun : 명사 (데이터, 분석, ...)
    - Verb : 동사 (하다, 먹다, ...)
    - Adjective : 형용사 (재미있다, 좋다, ...)
    - Josa : 조사 (은, 는, 이, 가)
    - Adverb : 부사 (정말, 매우)
    - KoreanParticle : 한국어 파티클 / 이모티콘 (예 : ㅋㅋㅋ, ㅠㅠ)
    - Punction : 구두점 (. , ? ~)
    - Number : 숫자
    - Alpha : 영문자
    - Modifier : 관형사 (새, 헌, 첫, ...)
    - Foreign : 외국어 / 한자
    - ScreenName : 트위터 아이디 표현 (@username)
    - Hashtag : 해시태그 (#태그)
- 매개변수
    - 정규화 (norm)
        - 문자를 정규화
        - 비표준어, 오타, 불필요한 반복 등을 깔끔하게 다듬어준다.
        - 문자들에거 노이즈를 제거
    - 어간 추출 (stem)
        - 사전에 등록된 기본 형태만 남기는 방법
        - 한국어의 동사와 형용사는 문맥에 따라서 형태가 매우 많다. 꼬리(어미)를 잘라내고, 기본형으로 통일

```python
from konlpy.tag import Okt
```

```python
okt=Okt()

#오류 발생 - 재부팅 헤보기!
```

```python
text = '파이썬에서 자바의 환경변수 설정은 너무 힘이 듭니다. ㅠㅠ'

#형태소 분리
tokens = okt.morphs(text)
print(tokens)
```

```python
text.split()
```

```python
text2='파이썬에서자바의환경변수설정은너무힘이듭니다.ㅠㅠ'
tokens2 = okt.morphs(text2)
print(tokens2)
```

```python
text2.split()
```

```python
#명사 추출
nouns=okt.nouns(text)
print(nouns)
```

```python
#픔사 태깅
#단어와 해당 단어의 품사를 튜플의 형태로 묶어서 반환
print(okt.pos(text))
```

```python
#Josa, Punctuation 품사를 제외하고 나머지 문자들만 사용
new_list=[]

for t, p in okt.pos(text):
	#t : 텍스트, p : 품사명
	#print(t)
	#print(p)
	#break
	if p not in ['Josa', 'Punctuation']:
		new_list.append(t)
new_list
```

```python
#정규화, 어간 추출
#norm : 문자의 오타, 비표준어, 불필요한 반복을 다듬어주는 부분
#stem : 꼬리(어미)를 잘라내고 기본형으로 변환
okt.pos(text, norm=True, stem=True)
```

### kkma 분석기
- 굉장히 깐깐한 언어 분석기
- 국어 사전에 나온 단어들을 이용하여 단어를 토큰환 하는 분석기

#### 세종 태그셋
- 명사류
	- NNG : 일반 명사 (데이터, 분석, ....)
	- NNP : 고유 명사 (서울, 컴퓨터, ...)
	- NNB : 의존 명사 (것, 수, 만큼, ..)
	- NP : 대명시 (나, 너 , 우리, ...)
- 동사 / 형용사
	- VV : 동사 (배우다, 먹다, ...)
	- VA : 형용사 (좋다, 빠르다, ...)
- 조사
	- JKS : 주격 조사 (이, 가)
	- JKO : 목적격 조사 (을, 를)
	- JKM : 부사격 조사 (에 ,에서, 으로)
- 기타
	- MAG : 일반 부사 (매우, 잘, 빨리)
	- EFM : 평서형 종결어미 (~다, ~는다)
	- SF : 마침표, 물음표, 느낌표

#### SentencePiece
- 문자의 덩어리를 학습하여 글자간의 상호 작용을 통해서 단어를 생성
- Train()을 이용하여 model을 생성하고 vocab 생성
    - input
        - 모델에서 학습할 문자들의 모음 (txt)
    - model_prefix
        - 모델의 이름을 지정
    - vocab_size
        - 단어 사전의 크기 (단어의 개수)
        - 8000, 160000, 320000을 일반적으로 사용
    - model_type
        - 토큰의 생성 방식
        - unigram : BERT, KoGPT 등에서 사용이 되는 언어 모델 방식
        - bpe : GPT-2에서 사용하는 방식 (한글에서는 부적합)
        - char : 문자 단위 (정보를 너무 잘게 쪼개는 형태)
        - word : 단어 단위 (한글에서는 부적합)
    - character_coverage
        - 학습에 포함할 문자의 종류 비율
        - 1.0을 사용하게 되면 모든 문자를 학습에서 사용
        - 0.9995는 한글, 영문, 숫자 포함
    - shuffle_input_sentence
        - 샘플링 시 문장의 순서를 섞어서 사용할 것인가?
        - 모델이 특정 순서에 편향 되는 것을 방지
    - input_sentence_size
        - 학습 문장을 샘플링
        - 모든 데이터를 사용하는 것이 가장 좋은 방법 (시간이 오래 걸림)
        - 일부만 샘플링하여 사용하는 방법
- 모델을 로드하고 서브워드로 문자를 나눠준다.

```python
spm.SentencePieceTrainer.Train(
    input='test.txt',
    model_prefix='ko_unigram',
    vocab_size=8000,
    model_type='unigram',
    character_coverage=0.9995,
    shuffle_input_sentence=True
)
```

```python
#모델을 로드
sp=spm.SentencePieceProcessor()

sp.load('ko_unigram.model')
```

```python
text4='나는 학교에 간다'
text5='스토리 개연성이 부족하지만 배우들의 연기는 뛰어났다'
```

```python
#단어 별로 데이터를 나눠주는 작업
print(sp.encode_as_pieces(text4))
print(sp.encode_as_pieces(text5))
```

```python
#class에서 문자열 클레스 변수 변경
#ClassMethod 데코레이터 함수
class Test:
    text='Hello'

    def __init__(self,a):
        self.a=a
    
    def func_1(self,x):
        Test.text=x
```

```python
class_1=Test(10)
class_2=Test(20)
```

```python
print(class_1.text)
print(class_2.text)
```

```python
#Class변수 text를 변경하는 함수를 호출
class_1.func_1('Hi')
```

```python
print(class_1.text)
print(class_2.text)
```

```python
class Test2:
    text='Hello'

    #일반적인 class 내부의 함수에서 첫번째 매개변수에는 생성이 된 class의 주소값이 입력(자동)
    def __init__(self,a):
        self.a=a
    
    #ClassMethod : 클래스의 이름에 접근할 수 있는 데코레이터 -> 함수에는 첫번째 매개변수에 클래스의 이름이 입력(자동)
    @classmethod
    def func_1(cls,x):
        #ClassMethod를 이용한 함수는 첫번째 매개변수 이름을 cls
        #cls에는 해당 class의 이름이 인자로 사용
        #수정 중에 class의 이름이 변하더라도 해당 cls는 그대로 유지가 가능
        
        cls.text=x
```

```python
class_3=Test2(10)
class_4=Test2(20)
```

```python
print(class_3.text)
print(class_4.text)
```

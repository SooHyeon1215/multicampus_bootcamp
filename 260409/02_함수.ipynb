### 문자열 내장 함수
- upper() : 문자열을 모두 대문자로 변환
- lower() : 문자열을 모두 소문자로 변환
- swapcase() : 대문자를 소문자로, 소문자를 대문자로 변환
- title() : 단어의 첫글자만 대문자로 변환
- strip() : 기본적으로는 문자열의 시작과 끝에 존재하는 공백 제거
    - lstrip() : 문자열 시작의 공백을 제거
    - rstrip() : 문자열 끝의 공백을 제거
- replace() : 특정한 문자를 찾아서 해당 문자를다른 문자로 치환
    - 문자 사이에 특정 문자를 치환(strip() 함수는 문자의 시작과 끝만 확인하고 replace()는 문자를 모두 확인)
- split() : 특정 문자를 찾아서 문자를 나눠주는 기능 -> 입력값은 문자, 츨력값은 list

```python
#sort() ->오름차순/내림차순 정렬을 할 수 있는 새로운 함수
#오름차순 정렬 : min()
#내림차순 정렬 : max()

#함수 선언
    #매개변수 2개
        #values 인자의 개수가 가변인 형태
        #reverse 기본값이 False(오름차순 정렬) // True (내림차순 정렬)
        
def custom_sort(*values,reverse=False):
    #정렬이 되는 데이터들을 저장할 공간(빈 리스트 생성)
    result=[]
    #values의 데이터가 수정을 할 수 있도록 list로 변환
    values = list(values)
    #values의 길이만큼 반복하는 반복문 생성 : values의 type이 tuple이기 때문
    for value in range(len(values)):
        #reverse의 값이 True라면 max()함수를 이용해서 values의 최대값을 추출
        if reverse:
            value=max(values)
        else:
            value=min(values)
        result.append(value)
        values.remove(value)
    return result
```

```python
custom_sort(3,2,4,1,reverse=True)
```

- types가 mean이라면 datas들의 평균 값을 되돌려준다.
- types가 위의 3가지 경우가 아니라면 'types 데이터 임력값이 잘못되었습니다.'라는 출력

```python
def func_12(*datas, types='max'):
    #결과 값 result 리스트 생성
    # result=[]
    #해당 함수 ->결과롤 돌려주는 데이터의 타입 -> 숫자(int, float) ->단일의 데이터
    #조건문 생성
    if types =='max':
        #datas에서 가장 큰 값을 반환 -> return 바로 사용 가능 // 변수를 생성하여 사장 큰 값을 대입
        #return max(datas)
        #return이 함수 안에서 실행되면 함수는 종료
        result=max(datas)
    elif types == 'min':
        #datas 에서 가장 작은 값을 되돌려준다.
        result = min(datas)
    elif types == 'mean':
        #datas의 평균 값을 되돌려준다.
        #평균 = 합계(sum // 반복문 이용) / 개수(len)
        res_sum = sum(datas)
        # res_sum_2 = 0
        # for data in datas:
        #     res_sum_2 += data
        result = res_sum/len(datas)
    else:
        #위의 세가지 조건이 모두 거짓인 경우
        print('types 데이터 입력값이 잘못되었습니다.')
        result='' # : 이유는?
    return result
```

```python
print(func_12(1,5,7)) # ->7
print(func_12(1,4,7,types='min')) # ->1
print(func_12(1,4,7,types='mean')) # ->4
print(func_12(1,4,7,types='sub')) # ->'type 데이터 입력값이 잘못되었습니다.'
```
